```python
import contractions

def expand_contractions(text):
    return contractions.fix(text)
```

```python
def clean_text(text):
    text = text.lower()

    text = re.sub(r'<.*?>', ' ', text)

    text = re.sub(r'http\S+|www\S+', ' ', text)

    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    text = re.sub(r'\s+', ' ', text).strip()

    return text
```

```python
df["expanded_text"] = df["review"].apply(expand_contractions)
```

```python
df["cleaned_text"] = df["expanded_text"].apply(clean_text)
```

```python
print(df["cleaned_text"].iloc[1])
```

```python
text = "I don't think this movie is good. It's not interesting."

print("Before:")
print(text)

print("\nAfter:")
print(expand_contractions(text))
```

```python
import nltk

nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words("english"))

print("Number of stop words:", len(stop_words))
print(list(stop_words)[:20])
```

```python
negation_words = {"no", "not", "never", "neither", "nor"}

stop_words = stop_words - negation_words
```

```python
def remove_stopwords(tokens):
    return [word for word in tokens if word.lower() not in stop_words]

df["filtered_tokens"] = df["tokens"].apply(remove_stopwords)
```

```python
print("Before stop-word removal:")
print(df["tokens"].iloc[0][:30])

print("\nAfter stop-word removal:")
print(df["filtered_tokens"].iloc[0][:30])
```

```python
import spacy

nlp = spacy.load("en_core_web_sm")
```

```python
df["filtered_text"] = df["filtered_tokens"].apply(lambda x: " ".join(x))
```

```python
df_sample = df.head(1000).copy()

print(df_sample.shape)
```

```python
df_sample["filtered_text"] = df_sample["filtered_tokens"].apply(
    lambda x: " ".join(x)
)
```

```python
lemmatized_results = []

for doc in nlp.pipe(df_sample["filtered_text"], batch_size=100):
    lemmatized_results.append([token.lemma_ for token in doc])

df_sample["lemmatized_tokens"] = lemmatized_results
```

```python
print("Before lemmatization:")
print(df_sample["filtered_tokens"].iloc[0][:30])

print("\nAfter lemmatization:")
print(df_sample["lemmatized_tokens"].iloc[0][:30])
```

```python
df_sample["preprocessed_text"] = df_sample["lemmatized_tokens"].apply(
    lambda x: " ".join(x)
)
```

```python
print("========== ORIGINAL ==========")
print(df_sample["review"].iloc[0])

print("\n========== EXPANDED ==========")
print(df_sample["expanded_text"].iloc[0])

print("\n========== CLEANED ==========")
print(df_sample["cleaned_text"].iloc[0])

print("\n========== TOKENS ==========")
print(df_sample["tokens"].iloc[0][:30])

print("\n========== AFTER STOP-WORD REMOVAL ==========")
print(df_sample["filtered_tokens"].iloc[0][:30])

print("\n========== LEMMATIZED ==========")
print(df_sample["lemmatized_tokens"].iloc[0][:30])

print("\n========== FINAL ==========")
print(df_sample["preprocessed_text"].iloc[0])
```

```python
comparison = pd.DataFrame({
    "Original Text": df_sample["review"].head(5),
    "Preprocessed Text": df_sample["preprocessed_text"].head(5)
})

display(comparison)
```

```python
df_sample["original_word_count"] = df_sample["review"].apply(
    lambda x: len(x.split())
)

df_sample["preprocessed_word_count"] = df_sample["preprocessed_text"].apply(
    lambda x: len(x.split())
)

display(
    df_sample[
        ["original_word_count", "preprocessed_word_count"]
    ].head(10)
)
```

```python
print("Final dataset shape:", df_sample.shape)

print("\nFinal columns:")
print(df_sample.columns.tolist())
```

```python
print("ORIGINAL:")
print(df_sample["review"].iloc[0][:500])

print("\n" + "="*80)

print("FINAL PREPROCESSED:")
print(df_sample["preprocessed_text"].iloc[0][:500])
```

```python
avg_original = df_sample["original_word_count"].mean()
avg_preprocessed = df_sample["preprocessed_word_count"].mean()

reduction = avg_original - avg_preprocessed
percentage = (reduction / avg_original) * 100

print("Average original word count:", round(avg_original, 2))
print("Average preprocessed word count:", round(avg_preprocessed, 2))
print("Average reduction:", round(reduction, 2))
print("Percentage reduction:", round(percentage, 2), "%")
```

The preprocessing pipeline was applied to a 1,000-review sample from the IMDb dataset. The average word count decreased after cleaning, stop-word removal, and lemmatization. Contraction expansion was performed before RegEx cleaning so contractions such as `don't` can be expanded correctly.

```python
print(word2vec_cbow.wv.most_similar("movie", topn=10))
```
